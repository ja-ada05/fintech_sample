```python
#fill all NA values with 0
application_overview.fillna(0, inplace=True)
application_overview.head(5)
```

```python
application_overview.head(5)

#plot the amt_income_total, grouped by income_type, and show the mean and median values for each group. Use a bar plot to visualize the results.
plt.figure(figsize=(10, 6))
sns.barplot(x='income_type', y='amt_income_total', data=application_overview, estimator=np.mean, color='skyblue', ci=None, label='Mean')
#include value on top of bar
sns.barplot(x='income_type', y='amt_income_total', data=application_overview, estimator=np.mean, color='skyblue', ci=None, label='Mean')
sns.barplot(x='income_type', y='amt_income_total', data=application_overview, estimator=np.median, color='navy', ci=None, label='Median')
plt.xlabel('Income Type')
plt.ylabel('Amount of Income')
plt.title('Mean and Median Amount of Income by Income Type')
plt.legend()
plt.show()
```
